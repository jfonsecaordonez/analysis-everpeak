
### 2.3 Corregir formatos numéricos y de fecha

**🎯Objetivo:**
Asegurar que las columnas de fechas y valores numéricos estén en formatos correctos para permitir análisis, cálculos y comparaciones precisas.

**Instrucciones:**

- Convierte las columnas de fecha de `traffic` a formato `datetime`. Haz el cambio a prueba de errores.
- En el dataset `eco`, limpia los valores numéricos:
    - En `city_gdp_capita`: elimina separadores de miles (`.`) y reemplaza las comas (`','`) por puntos (`'.'`) antes de convertir a tipo `float`.
    - En `unemployment_pct`: elimina el símbolo de porcentaje (`%`) y reemplaza las comas (`','`) por puntos (`'.'`) antes de convertir a tipo `float`.
    - En `population_m`: reemplaza las comas (`','`) por puntos (`'.'`) antes de convertir a tipo `float`.
- Finalmente, crea una nueva columna llamada `population` multiplicando `population_m` por 1,000,000 para obtener la población total.


```python
# Convertir las columnas de traffic a tipo fecha con pd.to_datetime()
traffic['update_time_utc']=pd.to_datetime(traffic['update_time_utc'])
traffic['update_time_utc_week_ago']= pd.to_datetime(traffic['update_time_utc_week_ago'])
# verificar el cambio
traffic.info()
```

<details>
<summary>Haz clic para ver la pista</summary>
para eliminar símbolos, puedes reemplazarlos por un texto vacío.

```python
#Limpia valores numéricos de city_gdp_capita
eco['city_gdp_capita']= (eco['city_gdp_capita']
    .astype('string')
    .str.strip()
    .str.replace('.','')
    .str.replace(',','.')
)
eco['city_gdp_capita']=pd.to_numeric(eco['city_gdp_capita'],errors='coerce')
#Limpia valores numéricos de unmployment_pct
eco['unemployment_pct']= (eco['unemployment_pct']
    .astype('string')
    .str.replace('%','')
    .str.strip()
    .str.replace(',','.')
)
eco['unemployment_pct']=pd.to_numeric(eco['unemployment_pct'],errors='coerce')

#Limpia valores numéricos de population_m
eco['population_m']= (eco['population_m']
    .astype('string')
    .str.strip()
    .str.replace(',','.')
)
eco['population_m']=pd.to_numeric(eco['population_m'],errors='coerce')

# Calcula la población total en unidades absolutas
eco['population_m']=eco['population_m']*1000000

# Verificar el cambio
eco.info()
eco.head(2)
```

```python

# Crear boxplot para observar el comportamiento de los minutos de congestion JamsDelay
# crea tu gráfico
sns.boxplot(
    x=traffic_2024_small['year'],
    y=traffic_2024_small['jams_delay'],
    showmeans=True
)
# obtener promedio para mostrarlo en título
mean_value = merged['jams_delay'].mean()
plt.title(f'Boxplot de JamsDelay (2024)\nPromedio: {mean_value:.2f}')

plt.show()
merged['jams_delay'].describe()
```

```python
# Crear histograma para ver la distribución de la economía (city_gdp_capita)
merged['city_gdp_capita'].hist(bins=5, figsize=(10, 5))

# obtener promedio para mostrarlo en el título
mean_value = merged['city_gdp_capita'].mean()

plt.title(f'Distribución del PIB per cápita (2024)\nPromedio: {mean_value:.2f}')
plt.xlabel('PIB per cápita')
plt.ylabel('Frecuencia (Número de Ciudades)')

plt.show()
```

RESUMEN EJECUTIVO

**Contexto & objetivo:**  
El análisis busca identificar la relación entre la movilidad urbana y la productividad económica, utilizando como variables principales el retraso total, el índice de tráfico, el tiempo de viaje por cada 10 km y el PIB per cápita. Estas variables representan indicadores clave, ya que el PIB per cápita permite aproximarse a la fortaleza económica de cada ciudad, mientras que el índice de tráfico, los retrasos y los tiempos de viaje permiten evaluar el comportamiento de la movilidad urbana. En conjunto, estos indicadores permiten analizar las condiciones económicas y de movilidad de las ciudades, proporcionando información útil para evaluar su atractivo para vivir e invertir. Sin embargo, las variables analizadas no presentan una correlación clara entre sí, por lo que no se evidencia que las ciudades con mayores niveles de congestión presenten, al mismo tiempo, menores niveles de productividad económica.



La base analizada contiene información de 15 ciudades de 2024 de América Latina. Se realizó una limpieza y estandarización de los datos, ajustando los tipos de variables, principalmente a float, y normalizando los nombres de las columnas en formato snake_case. Posteriormente, se integraron las variables de movilidad y economía mediante una unión INNER a nivel ciudad–año y se validaron los datos mediante distribuciones, posibles outliers y tendencias generales.



Los resultados muestran que Bogotá es un caso particularmente relevante, ya que presenta el mayor índice de tráfico de la muestra (37.61) y un PIB per cápita de US$11442.00, considerado intermedio frente a las demás ciudades. La relación entre tráfico y PIB per cápita es baja: Montevideo registra el mayor PIB (US$26176.0) y uno de los menores índices de tráfico (17.01), mientras que Ciudad de México presenta el segundo PIB más alto (US$21111.0) y uno de los mayores índices de tráfico. Además, se observa una posible relación entre el tráfico y el tamaño de la población, ya que varias de las ciudades con mayor congestión superan los 10 millones de habitantes.

Por otro lado, el boxplot de jams_delay mostró múltiples puntos por encima de la caja, lo que indica posibles valores atípicos asociados a ciudades con niveles de retraso superiores al comportamiento general de la muestra. Estos casos requieren una revisión adicional para identificar las ciudades involucradas y determinar si los mayores retrasos están relacionados con condiciones particulares de movilidad.

Por último, Bogotá debería considerarse una ciudad prioritaria para profundizar el análisis y evaluar inversiones en infraestructura de transporte, debido a la combinación de alta congestión, y productividad económica intermedia. No obstante, antes de establecer una relación causal, sería necesario incorporar más años de información, validar las fuentes y controlar factores como población, estructura económica y características de la red vial.
